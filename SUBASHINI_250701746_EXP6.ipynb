```python
import numpy as np
import pandas as pd

# Read the dataset
df = pd.read_excel('/content/pre-process_datasample.csv')

print(df)
print("")

# Fill missing value in Country with mode
df['Country'] = df['Country'].fillna(df['Country'].mode()[0])

# Separate features and label
features = df.iloc[:, :-1].values
label = df.iloc[:, -1].values

# Handle missing values in Age and Salary
from sklearn.impute import SimpleImputer

age = SimpleImputer(strategy="mean", missing_values=np.nan)
salary = SimpleImputer(strategy="mean", missing_values=np.nan)

age.fit(features[:, [1]])
salary.fit(features[:, [2]])

features[:, [1]] = age.transform(features[:, [1]])
features[:, [2]] = salary.transform(features[:, [2]])

print("After handling missing values:")
print(features)
print("")

# One Hot Encoding for Country
from sklearn.preprocessing import OneHotEncoder

oh = OneHotEncoder(sparse_output=False)

Country = oh.fit_transform(features[:, [0]])

print("One Hot Encoded Country:")
print(Country)
print("")

# Combine encoded Country with Age and Salary
final_set = np.concatenate(
    (Country, features[:, [1, 2]]),
    axis=1
)

print("Final Dataset:")
print(final_set)
print("")


from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

sc.fit(final_set)

feat_standard_scaler = sc.transform(final_set)

print("Standard Scaled Data:")
print(feat_standard_scaler)
print("")


from sklearn.preprocessing import MinMaxScaler

mms = MinMaxScaler(feature_range=(0, 1))

mms.fit(final_set)

feat_minmax_scaler = mms.transform(final_set)

print("Min-Max Scaled Data:")
print(feat_minmax_scaler)
```
