```python
# EXPERIMENT 6
# Handling Missing and Inappropriate Data in a Dataset

import numpy as np
import pandas as pd

# Upload Hotel.csv and convert it into DataFrame
# Make sure to upload the 'Hotel_Dataset.csv' file to your Colab environment
# If uploaded, it will typically be found in the /content/ directory.
df = pd.read_csv('Hotel_Dataset.csv')

# Display the dataset
df

# Identify duplicate rows
df.duplicated()

# Display information about the DataFrame
df.info()

# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Display the dataframe
df

# Find the number of rows
len(df)

# Reset the index
index = np.array(list(range(0, len(df))))
df.set_index(index, inplace=True)

# Display index
index

# Display dataframe
df

# Remove duplicate Age_Group column
df.drop(['Age_Group.1'], axis=1, inplace=True)

# Display dataframe
df
```

```python
# Replace negative CustomerID values with NaN
df.loc[df.CustomerID < 0, 'CustomerID'] = np.nan

# Replace negative Bill values with NaN
df.loc[df.Bill < 0, 'Bill'] = np.nan

# Replace negative EstimatedSalary values with NaN
df.loc[df.EstimatedSalary < 0, 'EstimatedSalary'] = np.nan

# Replace invalid NoOfPax values
# NoOfPax should be between 1 and 20
df.loc[(df['NoOfPax'] < 1) | (df['NoOfPax'] > 20), 'NoOfPax'] = np.nan

# Display dataframe
df
```

```python
# Display unique Age_Group values
df.Age_Group.unique()

# Display unique Hotel values
df.Hotel.unique()

# Replace incorrect hotel name
df.Hotel.replace(['Ibys'], 'Ibis', inplace=True)

# Display unique food preferences
df.FoodPreference.unique()

# Convert different forms of Veg to one standard value
df.FoodPreference.replace(
    ['Vegetarian', 'veg'],
    'Veg',
    inplace=True
)

# Convert different forms of Non-Veg to one standard value
df.FoodPreference.replace(
    ['non-Veg'],
    'Non-Veg',
    inplace=True
)

# Display dataframe
df
```

```python
# Fill missing EstimatedSalary using mean
df.EstimatedSalary.fillna(
    round(df.EstimatedSalary.mean()),
    inplace=True
)

# Fill missing NoOfPax using median
df.NoOfPax.fillna(
    round(df.NoOfPax.median()),
    inplace=True
)

# Fill missing Rating using median
df['Rating(1-5)'].fillna(
    round(df['Rating(1-5)'].median()),
    inplace=True
)

# Fill missing Bill using mean
df.Bill.fillna(
    round(df.Bill.mean()),
    inplace=True
)

# Display final dataframe
df
```
